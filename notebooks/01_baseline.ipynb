# 01 — Baseline & Zero-Shot Evaluation

This notebook evaluates off-the-shelf MegaDescriptor zero-shot embeddings on Population A (GZGC) and Population B (Grevy's) without additional fine-tuning.

```python
import torch
from zebraid.models.backbone import build_embedder
from zebraid.models.evaluate import compute_cmc_map
from zebraid.data.loaders import build_datasets
from zebraid.data.transforms import eval_transforms

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

transform = eval_transforms(384)
ds_a_test, ds_b_test = build_datasets('test', transform=transform)

model = build_embedder('megadescriptor', embedding_dim=512, pretrained=True, device=device)
model.eval()

print("Evaluating zero-shot MegaDescriptor on GZGC (Pop A)...")
metrics_a = compute_cmc_map(model, ds_a_test, device=device)
print(f"Pop A — Rank-1: {metrics_a['rank1']:.4f}, mAP: {metrics_a['map']:.4f}")

print("Evaluating zero-shot MegaDescriptor on Grevy's (Pop B)...")
metrics_b = compute_cmc_map(model, ds_b_test, device=device)
print(f"Pop B — Rank-1: {metrics_b['rank1']:.4f}, mAP: {metrics_b['map']:.4f}")
```
